```python
import time

# Reading the Inventory
try:
    fd = open('Inventory Management.txt', 'r')
    products = fd.read().strip().split('\n')  # Strip extra whitespace or blank lines
    fd.close()
except FileNotFoundError:
    print("Error: 'Inventory Management.txt' file not found.")
    exit()

# Taking User Input
ui_username = input("Enter your name: ")
ui_phone = input("Enter your Phone No: ")
ui_mail = input("Enter your Mail: ")
ui_prod_id = input("Enter product ID: ")
ui_prod_qn = int(input("Enter product Quantity: "))

updated_product_lst = []
product_found = False

# Going through each product details
for product in products:
    if product.strip() == "":  # Skip empty lines
        continue

    prod_details = product.split(',')

    if prod_details[0] == ui_prod_id:
        # Product ID matched
        product_found = True
        if int(ui_prod_qn) <= int(prod_details[3]):  # Check if enough quantity
            print("---------------------------------------")
            print("Product Name   : ", prod_details[1])
            print("Price          : ", prod_details[2])
            print("Quantity       : ", ui_prod_qn)
            print("---------------------------------------")
            print("Billing Amount : ", int(ui_prod_qn) * int(prod_details[2]))
            print("---------------------------------------")

            # Update inventory quantity
            prod_details[3] = str(int(prod_details[3]) - ui_prod_qn)

            # Generating sales in sale.txt
            fd = open("sale.txt", 'a')
            sale_detail = (
                ui_username + "," + ui_phone + "," + ui_mail + "," + prod_details[1] + ","
                + ui_prod_id + "," + str(ui_prod_qn) + "," 
                + str(int(ui_prod_qn) * int(prod_details[2])) + "," + time.ctime() + "\n"
            )
            fd.write(sale_detail)
            fd.close()

        else:
            print("Sorry, we are not having enough quantity.")
            print("We have only", prod_details[3], "quantity available.")
            choice = input("Would you like to purchase the available quantity? (Y/N): ").strip().lower()

            if choice == 'y':
                print("---------------------------------------")
                print("Product Name   : ", prod_details[1])
                print("Price          : ", prod_details[2])
                print("Quantity       : ", prod_details[3])
                print("---------------------------------------")
                print("Billing Amount : ", int(prod_details[3]) * int(prod_details[2]))
                print("---------------------------------------")

                # Update inventory to zero
                sale_quantity = prod_details[3]
                prod_details[3] = '0'

                # Generating sales in sale.txt
                fd = open("sale.txt", 'a')
                sale_detail = (
                    ui_username + "," + ui_phone + "," + ui_mail + "," + prod_details[1] + ","
                    + ui_prod_id + "," + sale_quantity + "," 
                    + str(int(sale_quantity) * int(prod_details[2])) + "," + time.ctime() + "\n"
                )
                fd.write(sale_detail)
                fd.close()
            else:
                print("Thanks for visiting!")

    # Append updated product details to the list
    updated_product_lst.append(prod_details)

if not product_found:
    print("Error: Product ID not found in inventory!")

# Writing back updated inventory to file
lst = []
for i in updated_product_lst:
    if len(i) >= 4:
        prod = ",".join(i) + '\n'
        lst.append(prod)

if lst:
    lst[-1] = lst[-1].strip()  # Remove the last newline character

fd = open('Inventory Management.txt', 'w')
fd.writelines(lst)
fd.close()

print("Inventory updated successfully!")
```
