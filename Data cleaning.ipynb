# Outlier
### An Outlier is an observation in a given dataset that lies far from the rest of the observations.

```python
data = pd.read_csv("Scholarship.csv")
```

```python
df = data.copy()
```

```python
df.head()
```

```python
df.shape
```

```python
plt.subplot(1,2,1)
sns.histplot(df["semester_percentage"])
plt.subplot(1,2,2)
sns.histplot(df["scholarship_exam_marks"])
plt.show()
```

```python
mean = df["semester_percentage"].mean()
std = df["semester_percentage"].std()
```

```python
print(mean,"," , std)
```

```python
Highest_allowed = mean + 3* std
Lowest_allowed = mean - 3* std
print(Highest_allowed, ",", Lowest_allowed)
```

```python
df[(df["semester_percentage"]< Lowest_allowed) | (df["semester_percentage"]> Highest_allowed)]
```

## Trimming

```python
df[(df["semester_percentage"]> Lowest_allowed) & (df["semester_percentage"]< Highest_allowed)]
```

```python
df["z_score"] = (df["semester_percentage"]-df["semester_percentage"].mean())/df["semester_percentage"].std()
```

```python
df.head()
```

```python
df[(df["z_score"]<-3) | (df["z_score"]>3)]
```

```python
df[(df["z_score"]>-3) & (df["z_score"]<3)]
```

## Capping

```python
upper_limit = mean + 3*std
lower_limit = mean - 3*std
```

```python
df["semester_percentage"] = np.where(df["semester_percentage"]> upper_limit,upper_limit,
                                     np.where(df["semester_percentage"]<lower_limit , lower_limit,df["semester_percentage"])
                                    )
```

```python
df
```

```python
df[(df["semester_percentage"]< Lowest_allowed) & (df["semester_percentage"]> Highest_allowed)]
```

## Quartiles:
These points which divide the total data into four equal parts when it has been orderly arranged are known as Quartiles.

## Interquartile Range [IQR]
The data range that contains the middle 50% of the entries.
IQR = Q3 - Q1


```python
from matplotlib import pyplot as plt
```

```python
data = pd.read_csv("scholarship.csv")
```

```python
data.head()
```

```python
sns.distplot(data["scholarship_exam_marks"])
```

```python
sns.boxplot(data["scholarship_exam_marks"])
```

```python
Q1 = data["scholarship_exam_marks"].quantile(0.25)
Q3 = data["scholarship_exam_marks"].quantile(0.75)
```

```python
Q1
```

```python
Q3
```

```python
IQR = Q3 - Q1
```

```python
data.describe()
```

```python
min = Q1 - 1.5*IQR
max = Q3 + 1.5*IQR
```

```python
min
```

```python
max
```

```python
data[(data["scholarship_exam_marks"]<min) | (data["scholarship_exam_marks"]>max) ]
```

## Trimming

```python
new_data = data[(data["scholarship_exam_marks"]>min) & (data["scholarship_exam_marks"]<max) ]
```

```python
plt.figure(figsize=(16,8))
plt.subplot(2,2,1)
sns.distplot(data["scholarship_exam_marks"])

plt.subplot(2,2,2)
sns.boxplot(data["scholarship_exam_marks"])

plt.subplot(2,2,3)
sns.distplot(new_data["scholarship_exam_marks"])

plt.subplot(2,2,4)
sns.boxplot(new_data["scholarship_exam_marks"])

plt.show()
```

## Capping 

```python
new_data2 = data.copy()
new_data2["scholarship_exam_marks"] = np.where(data["scholarship_exam_marks"]>max,
         max, 
         np.where(data["scholarship_exam_marks"]<min,min, data["scholarship_exam_marks"]))
```

```python
data.shape
```

```python
sns.boxplot(data["scholarship_exam_marks"])
```

## Outlier Removal Using Percentile
The percentile is a statistical measure that gives the percentage of data values that fall below a given data value

### How to check for outlier ?
DEcide percentage on both sides which you consider as outlier 
Example : we can take data which is below 2 percentile and above 98 percentile and consider them as outlier

```python
df = pd.read_csv("weight-height.csv")
```

```python
df.head()
```

```python
sns.displot(df["Height"])
```

```python
sns.boxplot(df["Height"])
```

```python
df.describe()
```

```python
min_limit = df["Height"].quantile(0.01)
max_limit = df["Height"].quantile(0.99)
```

```python
min_limit
```

```python
max_limit
```

```python
df[(df["Height"]>max_limit) | (df["Height"]<min_limit)]
```

```python
df.shape
```

## Trimming

```python
df_new = df[(df["Height"]<max_limit) & (df["Height"]>min_limit)]
```

```python
sns.boxplot(df_new["Height"])
```

## Capping

```python
df["Height"] = np.where(df["Height"]>max_limit , max_limit,
               np.where(df["Height"]<min_limit, min_limit, df["Height"]))
```

```python
df.shape
```

```python
sns.boxplot(df["Height"])
```
