A school claimed that the students study that is more intelligent than the average school. On calculating the IQ scores of 30 students, the average turns out to be 140. The mean of the population IQ is 100 and the standard deviation is 20 . State whether the claim of principle is right or not at a 5% significance level

```python
alpha = 0.05
```

```python
sample_mean = 140
sample_std = 20
data = sample_std*randn(30) + sample_mean
```

```python
print(np.mean(data),(np.std(data)))
```

```python
t_statistic,pvalue =stats.ttest_1samp(a = data, popmean = 100)
```

```python
if (pvalue<alpha):
    print("Reject the null hypothesis")
else:
    print("Failed to reject the null hypothesis")
```

### ANOVA

```python
from scipy.stats import f_oneway
```

Performance when each of the engine with different types of oil is applied

```python
performance1 = [89, 89, 88, 78, 79]
performance2 = [93, 92, 94, 89, 88]
performance3 = [89, 88, 89, 93, 90]
performance4 = [81, 78, 81, 92, 82]
```

```python
alpha = 0.05
```

```python
stat, pvalue = f_oneway(performance1, performance2, performance3, performance4)
```

```python
stat, pvalue
```

```python
if(pvalue<alpha):
    print("Reject the null Hypothesis")
else:
    print("Failed to reject the null Hypothesis")
```

This implies that we have sufficient proof to say that there exists a difference in the performance among four different engine oils.

```python
import statsmodels.api as sm 
from statsmodels.formula.api import ols
```

Test whether plant growth is affected by fertilizers and watering frequency.

```python
# Create a dataframe
dataframe = pd.DataFrame({'Fertilizer': np.repeat(['daily', 'weekly'], 15),
                          'Watering': np.repeat(['daily', 'weekly'], 15),
                          'height': [14, 16, 15, 15, 16, 13, 12, 11, 14, 
                                     15, 16, 16, 17, 18, 14, 13, 14, 14, 
                                     14, 15, 16, 16, 17, 18, 14, 13, 14, 
                                     14, 14, 15]})
```

```python
dataframe
```

```python
model = ols('height ~ Fertilizer + Watering + Fertilizer : Watering',  data=dataframe).fit()

result = sm.stats.anova_lm(model, type=2)

print(result)
```
